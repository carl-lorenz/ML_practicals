```python
import numpy as np
import pandas as pd
import sklearn.metrics as metrics
```

## Pre-processing

```python
def load_listings():
    """
    Removes listings without entry for target variable.
    Converts prices to float.
    :return: DataFrame: listings_df
    """
    listings_df = pd.read_csv("data/listings.csv")
    listings_df = listings_df.dropna(subset=["price"])
    listings_df = listings_df.reset_index(drop=True)
    listings_df["price"] = listings_df["price"].apply(lambda x: float(x.replace("$", "").replace(",", "")))
    return listings_df
```

## Training

```python
def validate(X_test, y_test, y_pred):
    std = y_test.std()
    rmse = np.sqrt(metrics.mean_squared_error(y_test, y_pred))
    mae = metrics.mean_absolute_error(y_test, y_pred)
    r2 = metrics.r2_score(y_test, y_pred)
    print(f"RMSE: {rmse} (std: {std})")
    print(f"MAE: {mae}")
    print(f"R2: {r2}")
```
